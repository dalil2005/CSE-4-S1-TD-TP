### **Step 3: Import the SimpleImputer**
```python
from sklearn.impute import SimpleImputer
```

This module provides multiple strategies to replace missing data:
- **mean** → replaces with the average (numeric only) 
- **median** → replaces with the median (recommended for numeric data with outliers) 
- **most_frequent** → replaces with the most common value (for categorical data) 
- **constant** → replaces with a fixed value (user-defined)

---

````python
from sklearn.impute import SimpleImputer
````

### **Step 4: Impute Numeric Data (e.g., `order_value_EUR`)**
```python
target_column = 'order_value_EUR'

# Create and apply the SimpleImputer using the median strategy
imputer = SimpleImputer(strategy='median')
df[target_column] = imputer.fit_transform(df[[target_column]])
```

 **Explanation:**
- `fit_transform()` → calculates the median and replaces missing values. 
- Double brackets `[[ ]]` ensure a 2D format required by scikit-learn. 
- After this operation, the numeric column no longer has missing data.

---

````python
target_column = 'order_value_EUR'

# Create a SimpleImputer instance to impute missing values with a strategy (e.g., 'mean', 'median', 'most_frequent')
# Other strategies include 'constant' to replace missing values with a constant value
imputer = SimpleImputer(strategy='median')

# Fit the imputer on the selected column and transform it to impute missing values
df[target_column] = imputer.fit_transform(df[[target_column]]).ravel()
# ravel(): flattens a multi-dimensional array into a one-dimensional array.
````

### Handling NaN Values with Conditional Replacement (loc method)

````python
# String column: fill with constant value (e.g., "Unknown")
df.loc[df['country'].isnull(), 'country'] = 'Algeria'
````

````python
#  Custom condition: fill NaN based on another column
df.loc[(df['cost'].isnull()) & (df['country'] == 'UK'), 'cost'] = 250000
````

## **Part 2 : Handling Non-Numeric and change types**

````python
# Sample variable containing mixed data
mixed_data = df['cost']

# Create an empty list to store non-numeric values
non_numeric_values = []

# Iterate through the elements in the variable
for value in mixed_data:
     if isinstance(value, str) and not value.isnumeric():
         non_numeric_values.append(value)

# Print the non-numeric values
print("Non-Numeric Values:")
print(non_numeric_values)
````

````python
# Remove this value

# df = df[df['cost'] != 'XXX'] 

df = df[~df['cost'].isin(non_numeric_values)] 
````

### Standardization of data types

````python
df['cost'] = df['cost'].astype(float) #convert type to float 
df['date'] = pd.to_datetime(df['date'])
````

## **Part 5 : Normalization and Standardization**

### MinMaxScaler --- Normalization

Scales values to a fixed range, usually **\[0, 1\]**.

**Formula:**\
$$ X' = \frac{X - X_{min}}{X_{max} - X_{min}} $$

**Example:**\
If `X = [10, 20, 30]`\
→ `X' = [0, 0.5, 1]`

------------------------------------------------------------------------

### StandardScaler --- Standardization

Centers data around **mean = 0** and scales to **standard deviation =
1**.

**Formula:**\
$$ Z = \frac{X - \mu}{\sigma} $$

**Example:**\
If `X = [10, 20, 30]`, Mean = 20, Std = 8.16\
→ `Z = [-1.22, 0, +1.22]`


````python
from sklearn.preprocessing import MinMaxScaler , StandardScaler

df['cost1'] = MinMaxScaler().fit_transform(df[['cost']])
df['cost2'] = StandardScaler().fit_transform(df[['cost']])
````
