## Partie 4 — Étude de la régularisation (L2, L1)

Nous allons étudier l'impact des différentes régularisations sur le modèle :

L2 (Ridge) : pénalise la somme des carrés des poids

L1 (Lasso) : pénalise la somme des valeurs absolues


```python
#  L2 Regularization (code complet)
def compute_cost_and_gradients_L2(X, y, w, b, lambda_):
     m = X.shape[0]
     z = np.dot(X, w) + b
     A = 1 / (1 + np.exp(-z))
     # Convertit y en un tableau numpy de forme (m, 1)
     y = y.values.reshape(-1, 1)
     cost = -(1/m) * np.sum(y*np.log(A+1e-8) + (1-y)*np.log(1-A+1e-8)) + (lambda_/(2*m))*np.sum(np.square(w))
     dw = (1/m) * np.dot(X.T, (A - y)) + (lambda_/m)*w
     db = (1/m) * np.sum(A - y)
     return cost, dw, db

```

```python
# Training (with l2 regularization)
lambda_l2 = 1.0
w_l2, b_l2 = initialize_parameters(n_features)
w_l2, b_l2, costs_l2 = gradient_descent(X_train, y_train, w_l2, b_l2, alpha=0.5, num_iters=2000, compute_fn=compute_cost_and_gradients_L2, lambda_=lambda_l2)

plt.plot(costs_l2)
plt.title('Cost over iterations (L2)')
plt.xlabel('Iteration (sampled)')
plt.ylabel('Cost')
plt.show()

y_pred_test_l2 = predict(X_test, w_l2, b_l2)
print('Test accuracy (L2):', accuracy_score(y_test, y_pred_test_l2))
print('Confusion matrix (L2):\\n', confusion_matrix(y_test, y_pred_test_l2))
```
